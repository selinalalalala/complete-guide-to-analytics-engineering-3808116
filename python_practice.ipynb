```python
columns_to_check = ['EmpID', 'OrderDate', 'OrderType', 'Quantity']

for index, row in df.iterrows():
    null_columns = [col for col in columns_to_check if pd.isnull(row[col])]
    if null_columns:
        print(f"第 {index} 行的以下欄位有空值: {null_columns}")
```
